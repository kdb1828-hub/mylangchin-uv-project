# 문제 1-1 : 기본 Chain 만들기 - AI 요리사

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()
PROVIDER = "groq"
CFG = {
    "groq": {"key": os.getenv("GROQ_API_KEY"), "url": "https://api.groq.com/openai/v1", "model": "openai/gpt-oss-120b"},
    "openai": {"key": os.getenv("OPENAI_API_KEY"), "url": None, "model": "gpt-4o-mini"},
}
c = CFG[PROVIDER]
if not c["key"]:
    raise RuntimeError(f".env 파일에 {PROVIDER.upper()}_API_KEY 가 없습니다.")

llm = ChatOpenAI(api_key=c["key"], base_url=c["url"], model=c["model"], temperature=0.7)

pt = PromptTemplate.from_template(
    "당신은 친절한 AI 요리사입니다.\n"
    "사용자가 가진 재료: {ingredients}\n\n"
    "이 재료로 만들 수 있는 요리를 하나 추천하고, 아래 형식을 정확히 지켜 한국어로 답하세요.\n\n"
    "{ingredients}로 만들 수 있는 요리를 추천드립니다!\n\n"
    "추천 요리: <요리 이름>\n"
    "재료: <필요한 재료 목록>\n"
    "조리법:\n"
    "1. ...\n"
    "2. ...\n"
    "3. ..."
)

chain = pt | llm | StrOutputParser()

ing = input("재료를 입력하세요 (예: 계란, 밥, 김치): ") or "계란, 밥, 김치"
try:
    print(chain.invoke({"ingredients": ing}))
except Exception as e:
    print("오류:", e)
```
